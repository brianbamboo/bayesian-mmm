## 1. Objective and Methodology

### Business Objective

FreshBite Foods wanted to understand how its four marketing channels contributed to revenue and how to allocate a quarterly marketing budget of \\\\$1,000,000 more effectively.

The analysis focused on two questions:

1. How much incremental revenue is associated with each marketing channel?
2. How could the quarterly budget be reallocated to maximize expected incremental revenue?

### Data Overview

The analysis used 156 weekly observations spanning approximately three years.

The dataset included:

- **Marketing investment:** TV, paid search, paid social, and display advertising spend.
- **Business controls:** Price index, promotional activity, distribution, and holiday periods.
- **Outcome variable:** Weekly revenue.

### Modeling Approach

A Bayesian Marketing Mix Model was developed using PyMC.

The model incorporated three important components:

- **Advertising carryover (Adstock):** Captures the lingering effects of advertising beyond the week in which spending occurs.
- **Diminishing returns (Saturation):** Captures the tendency for additional advertising investment to produce progressively smaller incremental returns.
- **Business controls:** Accounts for factors such as pricing, promotions, distribution, seasonality, and underlying trends that also influence revenue.

A Bayesian framework was used to estimate channel effectiveness while quantifying uncertainty in the model's parameters and predictions.

### Budget Optimization

Posterior estimates from the fitted model were used to simulate alternative quarterly budget allocations.

The optimization objective was to maximize expected incremental revenue across a 13-week quarter, subject to a total budget constraint of \\\\$1,000,000.

The analysis also evaluated alternative saturation assumptions to understand how sensitive the recommended allocation was to model specifications.

## 5. Uncertainty and Model Limitations

Although the Bayesian MMM produced useful directional insights, several limitations should be considered before using its results to guide actual marketing investment decisions.

### Model Diagnostics

The fitted model showed satisfactory convergence based on key sampling diagnostics, including R-hat values near 1.0 and adequate effective sample sizes.

However, good convergence does not guarantee that the model has correctly identified the underlying marketing effects.

The analysis revealed substantial uncertainty in several parameters, particularly advertising carryover (adstock) and channel contribution estimates.

### Parameter Recovery

The analysis used synthetic data with known underlying marketing effects, allowing estimated contributions to be compared against their true values.

The fitted model underestimated the average contribution of all four channels, with the largest discrepancies observed for TV and paid search.

This demonstrates that a model can converge successfully while still failing to recover the underlying parameters accurately.

### Saturation Assumptions and Sensitivity Analysis

The model treated each channel's EC50 saturation parameter as fixed rather than estimating it from the data.

To assess the impact of this assumption, three scenarios were evaluated:

- Low EC50: 20% below the baseline assumptions.
- Base EC50: Original EC50 assumptions.
- High EC50: 20% above the baseline assumptions.

The expected improvement from budget optimization remained relatively consistent:

| Scenario | Expected revenue improvement | Probability of outperforming baseline |
|---|---:|---:|
| Low EC50 | \\$22,201 | 72.3% |
| Base EC50 | \\$24,264 | 73.8% |
| High EC50 | \\$19,107 | 75.6% |

However, the recommended channel allocations changed considerably across scenarios. This indicates that the specific budget recommendation is sensitive to assumptions about advertising saturation.

This analysis represents scenario-based sensitivity testing rather than a full Bayesian estimate of EC50 uncertainty, since the posterior distributions of the other model parameters were held fixed.

### Additional Limitations

- **Observational data:** The model identifies statistical relationships between marketing activity and revenue but does not independently establish causality.
- **Parameter uncertainty:** Channel contributions and ROAS estimates have wide credible intervals.
- **Model specification:** Results depend on the selected functional forms, priors, control variables, and assumptions.
- **Revenue versus profit:** ROAS measures incremental revenue per advertising dollar, not profitability.
- **Optimization risk:** An optimizer can exploit weaknesses in estimated response curves, producing allocations that may not perform as expected in practice.

### Overall Assessment

The model provides a useful framework for evaluating marketing effectiveness and exploring alternative budget scenarios. However, its estimates should be treated as uncertain decision-support evidence rather than definitive measurements of causal channel performance.

As new marketing and revenue data become available, the model should be refitted and evaluated. Predictions should be compared with actual campaign outcomes to identify changes in channel performance and potential model misspecification.
